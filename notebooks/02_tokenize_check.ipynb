```python
import json
import pandas as pd
from underthesea import word_tokenize
from pyvi import ViTokenizer
```

```python
def compare_tokenizers(text):
    under_res=word_tokenize(text,format="text")
    pvi_res=ViTokenizer.tokenize(text)
    return under_res,pvi_res
```

```python
pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_rows', 100)

INPUT_CLEANED_FILE = '../data/processed/cleaned.jsonl'

sample_texts=[]

with open(INPUT_CLEANED_FILE,'r',encoding="utf8") as f:
    for i,text in enumerate(f):
        if i>=50:
            break
        
        try:
            data=json.loads(text)
            sample_texts.append(data.get('title', ''))
        except:
            continue
        
real_results=[]

for text in sample_texts:
    u_tok,p_tok=compare_tokenizers(text)
    is_match=(u_tok==p_tok)
    
    if not is_match:
        real_results.append({
            "Original Text (Cleaned)": text,
            "Underthesea": u_tok,
            "PyVi": p_tok
        })
        
df_real_diff=pd.DataFrame(real_results)
print(f"Phát hiện {len(df_real_diff)}/{len(sample_texts)} câu có sự khác biệt giữa 2 thư viện")
```
